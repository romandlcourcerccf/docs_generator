```python
from PIL import Image, ImageDraw, ImageFont
import os
from pathlib import Path
import time

def add_text_to_image(image_path, x, y, text, font_path=None, font_size=20, color=(255, 255, 255)):
    # Проверяем, существует ли файл
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"Файл {image_path} не найден.")

    # Открываем изображение
    img = Image.open(image_path)
    draw = ImageDraw.Draw(img)

    # Загружаем шрифт (если не указан — используем шрифт по умолчанию)
    try:
        if font_path:
            font = ImageFont.truetype(font_path, font_size)
        else:
            font = ImageFont.load_default()
    except Exception as e:
        print(f"Ошибка при загрузке шрифта: {e}. Используется шрифт по умолчанию.")
        font = ImageFont.load_default()

    # Рисуем текст
    draw.text((x, y), text, font=font, fill=color)

    # Сохраняем результат (в тот же файл)

    image_path = str(Path(image_path).parent)+ '/' + str(Path(image_path).stem) + time.strftime("%Y%m%d-%H%M%S") +'.' + str(Path(image_path).suffix)

    img.save(image_path)
    print(f"Текст успешно добавлен в {image_path}")


if __name__ == "__main__":
   
    add_text_to_image(
        image_path='/Users/roman/projects/docs_generator/files/inn.jpeg',
        x=400,
        y=780,
        # text=u'Иванов Иван Иванович',
        text =  u"Иванов Иван Иванович",
        # text=u'John Doe',
        font_path="/Users/roman/projects/docs_generator/fonts/Arial.ttf",
        font_size=40,
        color=(231, 76, 60)
    )
```
